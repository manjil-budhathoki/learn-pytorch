```python
import numpy as np
import torch
```

```python
data = [
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [5,6,7]
]
```

```python
np_data = np.array(data, dtype=np.float32)
print(np_data)
```

```python
torch_data = torch.from_numpy(np_data)
print(torch_data)
```

```python
print(torch_data.shape)
print(torch_data.ndim)
print(torch_data.numel())
print(torch_data.dtype)
```

```python
print(f"First row: {torch_data[0]}")
print(f"Second row: {torch_data[1]}")
print(f"Third row: {torch_data[2,2]}")
```
